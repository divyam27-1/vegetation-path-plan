```python
import simplekml
import json

fpath = "Data/infected.json"
with open(fpath) as f:
    infected = json.load(f)
```

```python
output_fpath = "Data/infected.kml"
docname = "Infected Zones"

kml = simplekml.Kml(name=docname)
kml.document.name = docname

for entry in infected:
    lat, lon = entry["coordinates"]
    val = entry["value"]

    pnt = kml.newpoint(name=f"{val:.2f}", coords=[(lon, lat)])
    pnt.style.iconstyle.color = simplekml.Color.red
    pnt.style.iconstyle.scale = 1.2

kml.save(output_fpath)
```
